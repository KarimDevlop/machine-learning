# Training Linear Regression

1. Solving for the weights

- g(X) = Xw
- Xw ~= y

- Suppose X is invertible

- -X⁻¹Xw = X⁻¹y
- -> Iw = X⁻¹y
- -> w = X⁻¹y

However X is in rectangular in our case (m rows and n+1 columns)
- -> The solution to this system simply doesn't exist.

2. The Gram matrix and the normal equation

Closest possible approximate solution: 
- Multiply both of Xw ~= y by the transpose of X: &emsp; <font color="green">XᵀXw = Xᵀy</font>
- The matrix XᵀX is called the gram matrix
- It has n+1 columns and n+1 rows
- We multiply both sides by (XᵀX)⁻¹: &emsp; <font color="green">(XᵀX)⁻¹XᵀXw = (XᵀX)⁻¹Xᵀy</font>
- Then we become this normal equation: &emsp; <font color="red">w = (XᵀX)⁻¹Xᵀy</font>

- This w is not the solution to the original system
- That solution doesn't exist
- But it is the closest possible solution.

3. Implementing the normal equation

```python
X = [
    [148, 24, 1385],
    [132, 25, 2031],
    [453, 11, 86],
    [158, 24, 185],
    [172, 25, 201],
    [413, 11, 86],
    [38,  54, 185],
    [142, 25, 431],
    [453, 31, 86],
]
X = np.array(X)
X
```

- First step of the normal equation is the Gram matrix:

```python
XTX = X.T.dot(X)
```

- Then we compute its inverse:

```python
XTX_inv = np.linalg.inv(XTX)
```

- Sanity-check that the inverse is correct:

```python
XTX_inv.dot(XTX).round(1)
```

- Target Vector y

```python
y = [100, 200, 150, 250, 100, 200, 150, 250, 120]
```

- Puitting all together, the normal equation gives us the weights:

```python
w_full = XTX_inv.dot(X.T).dot(y)
```

- We forgot the bias term
- To include it, we add a column of ones to X and repeat the same steps:

```python
ones = np.ones(X.shape[0])
# np.column_stack to add the ones column to the first column of X
X = np.column_stack([ones, X])
XTX = X.T.dot(X)
XTX_inv = np.linalg.inv(XTX)
w_full = XTX_inv.dot(X.T).dot(y)
```

- w_full can be split into two parts if we want

```python
w0 = w_full[0]
w = w_full[1:]
```

- The whole training in a function:

```python
def train_linear_regression(X, y):
    ones = np.ones(X.shape[0])
    X = np.column_stack([ones, X])

    XTX = X.T.dot(X)
    XTX_inv = np.linalg.inv(XTX)
    w_full = XTX_inv.dot(X.T).dot(y)

    return w_full[0], w_full[1:]
```
