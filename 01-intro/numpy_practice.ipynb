## 3. Multi-dimensional arrays

- indexing by row/column
- slices with `:`
- assign values to selected positions

```python
np.zeros((2, 3))
```

```python
np.ones((3,4))
```

```python
n = np.array([
    [1,20,3],
    [4,50,6],
    [7,8,90]
])
```

```python
n[0,1] = 2
n[1,1] = 5
n[2,2] = 9
n
```

```python
n[1] = [6,5,4]
n
```

```python
n[:,2]
```

```python
n[:,1] = [10,60,70]
n
```

```python
n[:,2] = [100, 600, 700]
n
```

## 4. Random arrays

- use `seed` for reproducibility

- `np.random.rand` for floats- `np.random.randint` for integers

```python
np.random.seed(42)
r = np.random.rand(4,3)
r
```

```python
100 * r
```

```python
np.random.seed(42)
np.random.randint(low=0, high=10, size=(3,4))
```

## 5. Element-wise operations

- works for arrays of same shape
- arithmetic is applied to each element

```python
a = np.arange(5)
a
```

```python
b = a**2 / 2
b
```

```python
c = b - a
c
```

## 6. Comparison operations

- boolean arrays can be used as masks
- comparisons return booleans

```python
b[2] > a[2]
```

```python
b > 2
```

```python
a > b
```

```python
a[a>b]
```

```python
a[a>=b]
```

```python
b[a<b]
```
